```python
import transformers
import evaluate
from datasets import load_dataset
```

```python
device = 'cuda' if torch.cuda.is_available() else 'cpu'
device
```

## Подготовка данных

```python
billsum = load_dataset('billsum', split='ca_test')
#ca_test - вид датасета (кол-во примеров)
```

```python
billsum
#text - текст
#summary - суммаризация
#title - название
```

```python
billsum = billsum.train_test_split(test_size=0.1)
```

```python
billsum
```

```python
tokenizer = transformers.AutoTokenizer.from_pretrained("ainize/bart-base-cnn")
```

```python
def preprocess_function(examples):
    model_inputs = tokenizer(examples['text'], max_length=1024, truncation=True)
    labels = tokenizer(text_target=examples['summary'], max_length=1024, truncation=True)

    model_inputs['labels'] = labels['input_ids']
    return model_inputs
```

```python
tokenized_billsum = billsum.map(preprocess_function, batched=True)
```

```python
model = transformers.AutoModelForSeq2SeqLM.from_pretrained("ainize/bart-base-cnn")
```

```python
data_collator = transformers.DataCollatorForSeq2Seq(tokenizer=tokenizer, model=model)
```

```python
training_args = transformers.Seq2SeqTrainingArguments(
    output_dir="./results",
    learning_rate=2e-5,
    per_device_train_batch_size=4,
    per_device_eval_batch_size=4,
    weight_decay=0.01,          # L2 регуляризация
    save_total_limit=3,
    num_train_epochs=2,
    report_to="none"
)
```

```python
tokenized_billsum
```

```python
trainer = transformers.Seq2SeqTrainer(
    model = model,
    args = training_args,
    train_dataset = tokenized_billsum['train'],
    eval_dataset = tokenized_billsum['test'],
    processing_class = tokenizer,
    data_collator = data_collator
)
```

```python
trainer.train()
```

## Предсказания на тестовом множестве

```python
# Предсказания на тестовом множестве
text_example = billsum['test']['text'][0]
print(text_example)
```

```python
input_ids = tokenizer.encode(
    text_example,
    return_tensors="pt",
    max_length=1024,
    truncation=True,
    ).to(device)
```

```python
input_ids.shape
```

```python
summary_text_ids = model.generate(input_ids = input_ids, max_length=142, min_length=56, num_beams=4,)
```

```python
summary_text_ids
```

```python
decoded_text = tokenizer.decode(summary_text_ids[0], skip_special_tokens=True)
```

```python
len(decoded_text), len(text_example)
```

```python
decoded_text
```

```python
summaries = []

for text in tqdm(billsum["test"]['text']):
    input_ids = tokenizer.encode(
        text,
        return_tensors="pt",
        max_length=1024,
        truncation=True,
        ).to(device)

    summary_text_ids = model.generate(
        input_ids=input_ids,
        bos_token_id=model.config.bos_token_id,
        eos_token_id=model.config.eos_token_id,
        max_length=142,
        min_length=56,
        num_beams=4,
    )

    decoded_text = tokenizer.decode(summary_text_ids[0], skip_special_tokens=True)
    summaries.append(decoded_text)
```

## Считаем качество

```python
!pip install rouge_score
```

```python
rouge = evaluate.load('rouge')
```

```python
%%time
results = rouge.compute(
        predictions=summaries,
        references=billsum["test"]['summary']
    )
```
